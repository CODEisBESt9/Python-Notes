# Conditional Statements

Conditional Statements in Python perform different computations or actions depending on whether a specific Boolean constraint evaluates to true or false. Conditional statements are handled by IF statements in Python.

In Python, we have four types of decision-making statements as given below:

• If Statement
• elif
• else
• If Else Statement
• Nested if Statement

- Rules

Decision making statements return Boolean value only. That is, either true or false.

Python programming language assumes any non-zero and non-null values as TRUE and zero or null values as FALSE values.


Syntax 

if condition:
    #code to execute, if this condition is true

else:
    #Code to execute, if the condition is false

# Only if

```python
num=199
if num<200:
    print('This number is less than 200') #condition true

print('Outside if block')
```

```python
num=200
if num<200:
    print('This number is less than 200')  #condition false

print('Outside if block')
```

# If......else

```python
a=25
b=52
if a>b:
    print('a is greater than b')

else:
    print('b is greater than a')  #condition is false, that why else condition execute

print('Outside the block')
```

```python
a=68
b=96

if a>b:
    print('a is greater than b')
elif b>a:
    print('b is greater than a')
else:
    print('Invalid')

print('Outside the block')
```

```python
age = int(input('Enter your age:- '))

if age >= 18:
    print('You are eligible to vote')
else:
    print('You are not eligible to vote')

print('Outside of block')
```

# Loops in Python are used to execute a block of code repetedly, baswd on a condition or  for a specified numbers of iterations, Python offers two main types of loops:

# for loop
The for loop is used for iterating over sequences (like lists, string, tuples or ranges). It automatically iterates through the elements of the sequence.  

```python
#Iterating over a range
for a in range(1,10,2):
    print(a)  #range(1,10,2) generates numbers from 1 to 9, with sonsecutive difference of 2 like 1,3,5,7,9
```

```python
for a in range(5):
    print(a) #it will generates from0 to 4, by default range start with 0.
```

```python
#iterating over a list(str)
list=['apple','banana','orange']
for g in list: #list itertaions
    print(g)
```

# while Loop
The while loop iterates or repeat the block of code until the condition is true

```python
#using a simple while loop
n=0
while (n<=10):
    print(n)
    n+=1
```

# Break statement: exits from the block of code or loop completely

```python
for i in range(10):
    if i==7:
        break  #When values of 'i' reaches to 7, it comes out from the loop
    print(i)
```

# Continue statement: skip the current iteration and moves to the next iteration.

```python
for i in range(10):
    if i==7:
        continue  #When values of 'i' reaches to 7, it comes out from the loop and continue from 8 (skip 7)
    print(i)
```

# Else clause: Works with both for and while loops and executes if the loop completes without a break.

```python
for i in range(7):
    print(i)
else:
    print('Loop Finished')
```

```python
for i in range(7):
    print(i)
    if i==7 :
        break  #here after the break, the else statement executed as another and last choice
else:
    print('Loop Finished')
```

```python
fruits = ['apple','banana','papaya']
for i in fruits:                         #we can directly print the elements of list using for or while loop
    print(i)
```

# Nested Loop

```python
#Outer Loop for rows
for row in range (1,4): #Rows : 1 to 3 
    #innner loop for columns
    for col in range(1,4): #columns 1 to 3
        print(f"({row}, {col})",end=' ') #Print the row and column
    print()  #Move to the next line
```
